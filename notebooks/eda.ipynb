```python
#figuring out the churned male customers by age
age_male_churned_one = df[(df['gender'] == 'Male') & (df['churn'] == 1)].groupby('age').size()
age_male_churned_zero = df[(df['gender'] == 'Male') & (df['churn'] == 0)].groupby('age').size()

#figuring out the churned female customers by age
age_female_churned_one = df[(df['gender'] == 'Female') & (df['churn'] == 1)].groupby('age').size()
age_female_churned_zero = df[(df['gender'] == 'Female') & (df['churn'] == 0)].groupby('age').size()

#creating a new dataframe to store the churned male and female customers by age
age_gender_churned = pd.DataFrame({
    'male_churned_one': age_male_churned_one,
    'male_churned_zero': age_male_churned_zero,
    'female_churned_one': age_female_churned_one,
    'female_churned_zero': age_female_churned_zero
})

age_gender_churned
```

```python
total_churned = age_gender_churned[age_gender_churned.index > 50].sum()

total_churned
```

-male and female customer seem to be pretty same, with male being slightly more dominant (5% more).

-female customer seem to leave the company slightly more than male customers.

-the stairs like trend seen in the gender churn rate (with churn being 0) seem to be explained by more male customers in those (and male customer staying in the company more than female customer)

-it seems male customer of 51 to 65 age cohort also leave nearly as much as female customer of that age cohorts with female customers having slightly more churned rate

```python
#figuring out total amount spend by different subscription types
total_spend = df['total_spend'].sum()
total_spend_basic = df[df['subscription_type'] == 'Basic']['total_spend'].sum()
total_spend_standard = df[df['subscription_type'] == 'Standard']['total_spend'].sum()
total_spend_premium = df[df['subscription_type'] == 'Premium']['total_spend'].sum()

print(f"Total spend: {total_spend}")
print(f"Total spend for Basic users: {total_spend_basic}")
print(f"Total spend for Standard users: {total_spend_standard}")
print(f"Total spend for Premium users: {total_spend_premium}")
```

data shows pretty similar churn rate for different subscription types and at the same time, each cohort seem to also spend similar amount.

```python
sns.histplot(data=df, x='contract_length', hue='churn')
```

```python
#figuring out total amount spend by different contract lengths
total_spend_con = df['total_spend'].sum()
total_spend_con_mon = df[df['contract_length'] == 'Monthly']['total_spend'].sum()
total_spend_con_quar = df[df['contract_length'] == 'Quarterly']['total_spend'].sum()
total_spend_con_ann = df[df['contract_length'] == 'Annual']['total_spend'].sum()

print(f"Total spend: {total_spend_con}")
print(f"Total spend for Monthly contracts: {total_spend_con_mon}")
print(f"Total spend for Quarterly contracts: {total_spend_con_quar}")
print(f"Total spend for Annual contracts: {total_spend_con_ann}")
```

data seem to show that monthly subscription shouldn't be priotized whereas annual and quarterly should be ideal subscription type.

possible solution could be offering discount offer if customer choose annual or quarterly contract

- churn rate seem to be nearly balanced with slightly more customer leaving the company

- there seem to be nearly equal male and female customers with male having slighter larger share (5% more). data also seem to suggest that female have higher churn rate than their male counterpart.

- customer with payment delay of more than 20 days should not be priotized to lower churn rate, as shown by data.
- data seem to suggest that it doesn't really matter the subscription type, the churn rate and their total spending is pretty similar.
- annual and quarterly contract length should be priotized more than monthly, as the data for churn rate and additional analysis shows.
- customer with total spending of 500 or more should be priotized as the data shows. it will not only help retain high spending customer but also lower the churn rate for this cohort, as churn rate for this cohort is lower.
- customer who interacted with the company in the last 15 days should be priotized over other customers, as the data shows lower churn rate for them.
